```python
import sympy as sp

from pylatex import NewLine


i_symb = sp.Symbol('i', complex=True)
x_symb = sp.Symbol('x')
X_symb = sp.Symbol('X')
k_symb = sp.Symbol('k')
N_symb = sp.Symbol('N')

sinc = sp.sinc(X_symb)
sinc_eqn = sp.sin(X_symb)/X_symb
partial_dx_sinc_uneval = sp.Derivative(sinc_eqn)
partial_dx_sinc = sp.diff(sinc_eqn)
sinc_dx_eqn_1_num = (sp.cos(X_symb) - sinc)
sinc_dx_eqn_1 = sinc_dx_eqn_1_num / X_symb

sinc_phase_priority_eqn_num = N_symb * i_symb**k_symb*(x_symb**(k_symb - 1))
sinc_phase_priority_eqn_denom = sp.gamma(k_symb + 1)
sinc_phase_priority_eqn = sinc_phase_priority_eqn_num / sinc_phase_priority_eqn_denom

latex_sinc = sp.latex(sinc)
latex_phase_priority_eqn = sp.latex(sinc_phase_priority_eqn)
latex_partial_dx_sinc = sp.latex(partial_dx_sinc)
latex_partial_dx_sinc_uneval = sp.latex(partial_dx_sinc_uneval)
latex_sinc_dx_eqn_1 = sp.latex(sinc_dx_eqn_1)
latex_sinc_dx_eqn_1_num = sp.latex(sinc_dx_eqn_1_num)

eqns = []

eqns.append('$$')

eqns.append(('$$', 'Optimising W.R.T frequency'))
eqns.append(latex_partial_dx_sinc_uneval)
eqns.append('=')
eqns.append(latex_partial_dx_sinc)
eqns.append('=')
eqns.append(latex_sinc_dx_eqn_1)
eqns.append((False, NewLine()))
eqns.append('\\implies X\\mathcal{E}(X) = ')
eqns.append(latex_sinc_dx_eqn_1_num)
eqns.append('\\implies \\cos(X) - X\\mathcal{E}(X) = ')
eqns.append(latex_sinc)

eqns.append(('$$', 'Optimising W.R.T phase'))
eqns.append((False, NewLine()))
eqns.append('\\mathcal{O}_{\\theta}(X) = ')
eqns.append(latex_phase_priority_eqn)

eqns.append('$$')
```
